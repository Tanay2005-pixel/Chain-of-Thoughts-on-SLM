# AQuA-RAT
## train dataset
## features:
    questions
    options
    rationals
    correct ans

## number of rows : 97,467

```python
import json

aqua_train = []

with open("AQuA/train.json", "r", encoding="utf-8") as f:
    for line in f:
        aqua_train.append(json.loads(line))

print("Total examples:", len(aqua_train))
print(aqua_train[0])
```

```python
print("THIS IS THE TRAIN DATASET OF THE AQuA-RAT", end="\n\n")

for i in range(3):
    print("QUESTION :", aqua_train[i]["question"])
    print("OPTION A :", aqua_train[i]["options"][0])
    print("OPTION B :", aqua_train[i]["options"][1])
    print("OPTION C :", aqua_train[i]["options"][2])
    print("OPTION D :", aqua_train[i]["options"][3])
    print("OPTION E :", aqua_train[i]["options"][4])
    print("Rationale :", aqua_train[i]["rationale"])
    print("ANSWER :", aqua_train[i]["correct"], end="\n\n")
```

## converting the dataset into the needed json file (removing rationals)

```python
converted_data = []

for i, row in enumerate(aqua_train):

    options = row["options"]

    # A → 0, B → 1, C → 2, D → 3, E → 4
    correct_index = ord(row["correct"].upper()) - ord("A")

    converted_data.append({
        "id": i + 1,
        "question": row["question"],
        "options": options,
        "answer": options[correct_index]
    })

# Save JSON
with open("aqua_train_converted.json", "w", encoding="utf-8") as f:
    json.dump(converted_data, f, indent=2, ensure_ascii=False)

print("Dataset converted and saved successfully!")
print("Total examples:", len(converted_data))
```

```python

import json
import os

input_file = "AQuA/train.json"
output_file = "aquarat_train_converted.json"

with open(input_file, "r", encoding="utf-8") as f:
    content = f.read().strip()

# Support both a standard JSON array and JSONL
try:
    data = json.loads(content)

except json.JSONDecodeError:
    data = [
        json.loads(line)
        for line in content.splitlines()
        if line.strip()
    ]

# Convert into the required format
converted_data = []

for i, item in enumerate(data, start=1):
    converted_item = {
        "id": i,
        "question": item["question"],
        "option": item["options"],
        "reasoning": item["rationale"],
        "ans": item["correct"]
    }

    converted_data.append(converted_item)

# Save as a new JSON file
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(
        converted_data,
        f,
        indent=2,
        ensure_ascii=False
    )

print("Conversion successful!")
print("Total questions:", len(converted_data))
print("Output file:", os.path.abspath(output_file))

# Preview first record
if converted_data:
    print(json.dumps(converted_data[0], indent=2, ensure_ascii=False))
```
