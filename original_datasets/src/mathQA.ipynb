```python
import re
import ast

def parse_options(options):

    # If it's a string representation of a Python list
    if isinstance(options, str) and options.strip().startswith("["):
        options = ast.literal_eval(options)

    # If we now have a list
    if isinstance(options, list):
        return [
            option.split(")", 1)[1].strip()
            for option in options
        ]

    # Otherwise it's the normal MathQA string format
    pattern = r"[a-e]\s*\)\s*(.*?)(?=\s*,\s*[a-e]\s*\)|$)"
    return re.findall(pattern, options)
```

```python
for i in [0, 63]:
    row = mathqa_train_dataset[i]

    choices = parse_options(row["options"])

    correct_index = ord(row["correct"]) - ord("a")
    correct_answer = choices[correct_index]

    print("INDEX:", i)
    print("CHOICES:", choices)
    print("CORRECT:", correct_answer)
    print()
```

```python
for i, row in enumerate(mathqa_train_dataset):

    choices = parse_options(row["options"])
    correct_index = ord(row["correct"]) - ord("a")

    if len(choices) != 5:
        print("Wrong number of choices:", i, choices)

    if correct_index >= len(choices):
        print("Invalid correct index:", i, row["correct"], choices)
```

```python
mathqa_validation_dataset = mathqa_data["validation"]
mathqa_test_dataset = mathqa_data["test"]
```

```python
for dataset_name, dataset in [
    ("validation", mathqa_validation_dataset),
    ("test", mathqa_test_dataset)
]:
    print("Checking:", dataset_name)

    for i, row in enumerate(dataset):
        choices = parse_options(row["options"])
        correct_index = ord(row["correct"]) - ord("a")

        if len(choices) != 5:
            print("Wrong number of choices:", i, choices)

        if correct_index >= len(choices):
            print("Invalid correct index:", i, row["correct"], choices)

    print("Done:", dataset_name)
```

```python
import os
import json

os.makedirs("MathQA", exist_ok=True)

def convert_mathqa(dataset):
    converted_data = []

    for i, row in enumerate(dataset):
        choices = parse_options(row["options"])
        correct_index = ord(row["correct"]) - ord("a")
        correct_answer = choices[correct_index]

        converted_data.append({
            "id": i + 1,
            "context": None,
            "question": row["Problem"],
            "choices": choices,
            "correct": correct_answer
        })

    return converted_data


# Convert all splits
mathqa_train = convert_mathqa(mathqa_train_dataset)
mathqa_validation = convert_mathqa(mathqa_validation_dataset)
mathqa_test = convert_mathqa(mathqa_test_dataset)


# Save files
with open("MathQA/train.json", "w", encoding="utf-8") as f:
    json.dump(mathqa_train, f, indent=2, ensure_ascii=False)

with open("MathQA/validation.json", "w", encoding="utf-8") as f:
    json.dump(mathqa_validation, f, indent=2, ensure_ascii=False)

with open("MathQA/test.json", "w", encoding="utf-8") as f:
    json.dump(mathqa_test, f, indent=2, ensure_ascii=False)

print("MathQA conversion complete!")
print("Train:", len(mathqa_train))
print("Validation:", len(mathqa_validation))
print("Test:", len(mathqa_test))
```
