```python
import json

# Load train.json
with open("strategyqa/data/strategyqa/train.json", "r", encoding="utf-8") as f:
    train_data = json.load(f)

# Load dev.json
with open("strategyqa/data/strategyqa/dev.json", "r", encoding="utf-8") as f:
    dev_data = json.load(f)

# Combine train + dev
all_data = train_data + dev_data

converted_data = []

for i, row in enumerate(all_data):

    # Combine term and description into context
    context = row["term"] + " " + row["description"]

    converted_data.append({
        "id": i + 1,
        "context": context,
        "question": row["question"],
        "answer": row["answer"]
    })

# Save combined JSON
with open("strategyqa_combined.json", "w", encoding="utf-8") as f:
    json.dump(converted_data, f, indent=2, ensure_ascii=False)

print("Successfully created strategyqa_combined.json")
print("Total examples:", len(converted_data))
```
