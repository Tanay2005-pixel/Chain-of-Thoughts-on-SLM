```python
from datasets import load_dataset
obqa_data = load_dataset("allenai/openbookqa", "main")

print(obqa_data)
```

```python
obqa_train = obqa_data["train"]

print(obqa_train)
print(obqa_train[0])
```

```python
print("THIS IS THE TRAIN DATASET OF THE OBQA",end="\n\n")
for i in range(len(obqa_train)):
  print("QUESTION :", obqa_train[i]["question_stem"])
  print("OPTION A :", obqa_train[i]["choices"]["text"][0])
  print("OPTION B :", obqa_train[i]["choices"]["text"][1])
  print("OPTION C :", obqa_train[i]["choices"]["text"][2])
  print("OPTION D :", obqa_train[i]["choices"]["text"][3])
  
  print("ANSWER :", obqa_train[i]["answerKey"],end = "\n\n")
```

```python
import json
converted_data = []

for i in range(len(obqa_train)):

    choices = obqa_train[i]["choices"]["text"]
    answer_key = obqa_train[i]["answerKey"]

    correct_index = ord(answer_key) - ord("A")

    converted_data.append({
        "id": i + 1,
        "question": obqa_train[i]["question_stem"],
        "options": choices,
        "answer": choices[correct_index]
    })

# Save directly in the pod's current Jupyter folder
with open("obqa_train.json", "w", encoding="utf-8") as f:
    json.dump(converted_data, f, indent=2, ensure_ascii=False)

print("Saved successfully!")
print("Total examples:", len(converted_data))
```
