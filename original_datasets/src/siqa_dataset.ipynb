```python
from datasets import load_dataset
import pandas as pd
import json
```

# SIQA TRAINING DATASET

### FEATURES : 
    'context'
    'questions'
    'answerA'
    'answerB'
    'answerC'
    'label'

### number of rows : 33410

```python
siqa_data = load_dataset("lighteval/siqa")
```

```python
siqa_train_dataset = siqa_data["train"]
print(siqa_train_dataset)
type(siqa_train_dataset)
```

```python
print("THIS IS THE TRAIN DATASET OF THE SIQA",end="\n\n")
for i in range(3):
  print("CONTEXT :", siqa_train_dataset[i]["context"])
  print("QUESTION :", siqa_train_dataset[i]["question"])
  print("OPTION A :", siqa_train_dataset[i]["answerA"])
  print("OPTION B :", siqa_train_dataset[i]["answerB"])
  print("OPTION c :", siqa_train_dataset[i]["answerC"])
  print("ANSWER :", siqa_train_dataset[i]["label"],end = "\n\n")
```

## LOADING THE JSON FILE OF SIQA TRAINING DATASET

```python
converted_data = []

for i, row in enumerate(siqa_train_dataset):

    choices = [
        row["answerA"],
        row["answerB"],
        row["answerC"]
    ]

    # SIQA label is 1, 2, or 3
    correct_index = int(row["label"]) - 1

    converted_data.append({
        "id": i + 1,
        "context": row["context"],
        "question": row["question"],
        "choices": choices,
        "correct": choices[correct_index]
    })

# Save in the current Jupyter folder
file_path = "siqa_train.json"

with open(file_path, "w", encoding="utf-8") as f:
    json.dump(converted_data, f, indent=2, ensure_ascii=False)
```
